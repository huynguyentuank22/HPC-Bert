```python
import os
import pandas as pd
from sklearn.metrics import r2_score

input_folder = "predict_results"
output_folder = "baseline_results"

os.makedirs(output_folder, exist_ok=True)

for filename in os.listdir(input_folder):
    if not filename.endswith("_pred.csv"):
        continue

    file_path = os.path.join(input_folder, filename)

    # Đọc file predict + y_true
    df = pd.read_csv(file_path)

    # Giả sử cột tên: "y_true" và "y_pred"
    if "y_true" not in df.columns or "y_pred" not in df.columns:
        print(f"⚠️ File {filename} thiếu cột cần thiết. Bỏ qua.")
        continue

    y_true = df["y_true"]
    y_pred = df["y_pred"]

    # Tính R²
    r2 = r2_score(y_true, y_pred)

    # Tạo nội dung txt
    txt_content = f"R2: {r2:.4f}"

    # Tạo file output .txt
    out_name = filename.replace("_pred.csv", "_r2.txt")
    out_path = os.path.join(output_folder, out_name)

    # Lưu file txt
    with open(out_path, "w", encoding="utf-8") as f:
        f.write(txt_content)

    print(f"✅ R2 score saved: {out_name} ({r2:.4f})")
```

```python
import os

folder = "baseline_results"

for filename in os.listdir(folder):
    if not filename.endswith("_r2.txt"):
        continue

    r2_path = os.path.join(folder, filename)

    # tên file gốc (loại bỏ _r2)
    base_name = filename.replace("_r2.txt", ".txt")
    base_path = os.path.join(folder, base_name)

    # Nếu file gốc KHÔNG tồn tại → bỏ qua
    if not os.path.exists(base_path):
        print(f"⚠️ Không tìm thấy file gốc: {base_name}. Bỏ qua.")
        continue

    with open(base_path, "r", encoding="utf-8") as f:
        base_content = f.read()
    if "R2" in base_content:
        continue
    
    # Đọc nội dung R2
    with open(r2_path, "r", encoding="utf-8") as f:
        r2_content = f.read()

    # Append vào file gốc
    with open(base_path, "a", encoding="utf-8") as f:
        f.write("\n")
        f.write(r2_content)

    print(f"✅ Đã append R2 vào: {base_name}")
```
